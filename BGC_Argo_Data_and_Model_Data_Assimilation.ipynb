## What is actually happening beneath the ocean surface?

Imagine that somebody asks:

> <span style="color:blue;"><b>What is happening to phytoplankton in this part of the ocean today</b></span>

A satellite looks down from space and sees the 
<span style="color:blue;"><b>surface</b></span>.

A BGC-Argo float dives through the ocean and sees the <span style="color:blue;"><b>vertical structure</b></span>, but only when and where it profiles (records observations).

A numerical model attempts to connect those separate pieces into a continuous picture through <span style="color:blue;"><b>time and depth</b></span>.

But, none of these views is the whole ocean.

So,

> <span style="color:blue;"><b>Different observing and modelling tools see different parts of the same system.</b></span>.

In this practical session, we will gradually combine these pieces together and ask what each component adds value to our understanding of ocean.

## 5. Third view: <span style="color:blue;"><b>what does BGC-Argo see below the surface?</b></span>.

The profiling float gives us something the satellite cannot:

> **chlorophyll structure beneath the ocean surface.**

To make the individual profiles easier to interpret, we interpolate each observed profile onto a common depth grid and display them through time.

```python
argo_time = pd.to_datetime(
    ds["argo_time"].values
)

argo_depth = ds[
    "argo_depth"
].values

argo_chl = ds[
    "argo_chlorophyll"
].values


# --------------------------------------------------
# Identify the individual BGC-Argo profile dates
# --------------------------------------------------

profile_dates = pd.DatetimeIndex(
    argo_time
).normalize().unique()

normalised_time = pd.DatetimeIndex(
    argo_time
).normalize()


# --------------------------------------------------
# Common depth grid
# --------------------------------------------------

depth_grid = np.linspace(
    -200,
    0,
    101
)


# --------------------------------------------------
# Create a DAILY time axis.
#
# Days without a BGC-Argo profile remain NaN,
# so the observational gaps become visible.
# --------------------------------------------------

daily_dates = pd.date_range(
    profile_dates.min(),
    profile_dates.max(),
    freq="D"
)

argo_daily = np.full(
    (
        len(daily_dates),
        len(depth_grid)
    ),
    np.nan
)


# --------------------------------------------------
# Interpolate each observed profile vertically,
# but place it ONLY on the day it was observed.
# --------------------------------------------------

for date in profile_dates:

    mask = (
        normalised_time == date
    )

    d = argo_depth[mask]
    c = argo_chl[mask]

    good = (
        np.isfinite(d)
        & np.isfinite(c)
    )

    d = d[good]
    c = c[good]

    if len(d) < 2:
        continue

    order = np.argsort(d)

    d = d[order]
    c = c[order]

    d_unique, unique_index = np.unique(
        d,
        return_index=True
    )

    c_unique = c[
        unique_index
    ]

    inside = (
        (depth_grid >= d_unique.min())
        & (depth_grid <= d_unique.max())
    )

    profile_interp = np.full(
        len(depth_grid),
        np.nan
    )

    profile_interp[inside] = np.interp(
        depth_grid[inside],
        d_unique,
        c_unique
    )

    day_index = daily_dates.get_loc(
        pd.Timestamp(date)
    )

    argo_daily[
        day_index,
        :
    ] = profile_interp


# --------------------------------------------------
# Colour scale
# --------------------------------------------------

vmax = np.nanpercentile(
    argo_daily,
    98
)

cmap = plt.cm.viridis.copy()

# White = no BGC-Argo observation
cmap.set_bad("white")


# --------------------------------------------------
# Plot
# --------------------------------------------------

fig, ax = plt.subplots(
    figsize=(13, 6)
)

pcm = ax.pcolormesh(
    daily_dates,
    depth_grid,
    argo_daily.T,
    shading="nearest",
    cmap=cmap,
    vmin=0,
    vmax=vmax
)

ax.set_title(
    "BGC-Argo observations: chlorophyll through time and depth"
)

ax.set_xlabel("Time")
ax.set_ylabel("Depth (m)")
ax.set_ylim(-200, 0)


# Make date labels readable
ax.tick_params(
    axis="x",
    labelrotation=35
)

for label in ax.get_xticklabels():
    label.set_horizontalalignment(
        "right"
    )


cbar = plt.colorbar(
    pcm,
    ax=ax,
    pad=0.02
)

cbar.set_label(
    "Chlorophyll-a (mg m$^{-3}$)"
)

plt.tight_layout()
plt.show()


print(
    f"BGC-Argo profile dates : "
    f"{len(profile_dates)}"
)

print(
    f"Days in observation period: "
    f"{len(daily_dates)}"
)

print(
    f"Days with no BGC-Argo profile: "
    f"{len(daily_dates) - len(profile_dates)}"
)
```

### Same BGC-Argo observations, showing only the sampled profiles

```python
# --------------------------------------------------
# Same BGC-Argo data, but show only sampled profiles
# --------------------------------------------------

argo_grid_profiles = np.full(
    (
        len(profile_dates),
        len(depth_grid)
    ),
    np.nan
)

for i, date in enumerate(profile_dates):

    mask = (
        normalised_time == date
    )

    d = argo_depth[mask]
    c = argo_chl[mask]

    good = (
        np.isfinite(d)
        & np.isfinite(c)
    )

    d = d[good]
    c = c[good]

    if len(d) < 2:
        continue

    order = np.argsort(d)

    d = d[order]
    c = c[order]

    d_unique, unique_index = np.unique(
        d,
        return_index=True
    )

    c_unique = c[
        unique_index
    ]

    inside = (
        (depth_grid >= d_unique.min())
        & (depth_grid <= d_unique.max())
    )

    argo_grid_profiles[
        i,
        inside
    ] = np.interp(
        depth_grid[inside],
        d_unique,
        c_unique
    )


# --------------------------------------------------
# Colour scale
# --------------------------------------------------

vmax_profiles = np.nanpercentile(
    argo_grid_profiles,
    98
)


# --------------------------------------------------
# Plot only the sampled profile dates
# --------------------------------------------------

fig, ax = plt.subplots(
    figsize=(13, 6)
)

pcm = ax.pcolormesh(
    profile_dates,
    depth_grid,
    argo_grid_profiles.T,
    shading="auto",
    cmap="viridis",
    vmin=0,
    vmax=vmax_profiles
)

ax.set_title(
    "BGC-Argo chlorophyll: sampled profiles only"
)

ax.set_xlabel("Profile date")
ax.set_ylabel("Depth (m)")
ax.set_ylim(-200, 0)

ax.tick_params(
    axis="x",
    labelrotation=35
)

for label in ax.get_xticklabels():
    label.set_horizontalalignment(
        "right"
    )

cbar = plt.colorbar(
    pcm,
    ax=ax,
    pad=0.02
)

cbar.set_label(
    "Chlorophyll-a (mg m$^{-3}$)"
)

plt.tight_layout()
plt.show()
```

## 7. Combining the model and observations

We now introduce the basic idea of **data assimilation**.

The numerical model gives us one estimate.

The observation gives us another.

So the first question is:

> **Which one should we trust more?**

Lets look at this equation:
$$
x_a = x_m + K(x_o - x_m)
$$

where:

- $x_m$ = **model estimate**
- $x_o$ = **observation**
- $x_a$ = **updated (analysis) estimate**
- $K$ = **how strongly we move toward the observation**

What does $K$ mean?

- $K = 0$ → trust the **model**
- $K = 1$ → trust the **observation**
- $0 < K < 1$ → combine both

Before using the real BGC-Argo profile, let us explore this idea interactively.

They disagree.

```python
import ipywidgets as widgets
from IPython.display import display

# --------------------------------------------------
# Simple artificial example
# --------------------------------------------------

depth_demo = np.array([
    -150, -125, -100, -75, -50, -25, 0
])

model_demo = np.array([
    0.03, 0.04, 0.05, 0.07, 0.10, 0.12, 0.10
])

obs_demo = np.array([
    0.02, 0.05, 0.12, 0.18, 0.14, 0.09, 0.07
])


def show_assimilation(K=0.5):

    updated_demo = (
        model_demo
        +
        K * (
            obs_demo
            -
            model_demo
        )
    )

    fig, ax = plt.subplots(
        figsize=(7, 7)
    )

    ax.plot(
        model_demo,
        depth_demo,
        color=MODEL_COLOUR,
        linewidth=3,
        label="Model"
    )

    ax.scatter(
        obs_demo,
        depth_demo,
        color=OBS_COLOUR,
        s=70,
        label="Observation",
        zorder=5
    )

    ax.plot(
        updated_demo,
        depth_demo,
        color=UPDATE_COLOUR,
        linewidth=3,
        label="Updated estimate"
    )

    ax.set_xlabel(
        "Chlorophyll-a"
    )

    ax.set_ylabel(
        "Depth (m)"
    )

    ax.set_xlim(
        0,
        0.22
    )

    ax.set_title(
        f"How much should we trust the observation?   K = {K:.2f}"
    )

    ax.legend()

    plt.tight_layout()
    plt.show()

    if K < 0.3:
        message = "Mostly trusting the MODEL"

    elif K > 0.7:
        message = "Mostly trusting the OBSERVATION"

    else:
        message = "Combining MODEL and OBSERVATION"

    display(
        widgets.HTML(
            value=f"""
            <div style="
                font-size:22px;
                font-weight:bold;
                text-align:center;
                padding:10px;
            ">
                {message}
            </div>
            """
        )
    )


K_slider = widgets.FloatSlider(
    value=0.5,
    min=0.0,
    max=1.0,
    step=0.05,
    description="K:",
    continuous_update=True,
    readout_format=".2f",
    style={
        "description_width": "initial"
    },
    layout=widgets.Layout(
        width="75%"
    )
)

display(
    widgets.HTML(
        value="""
        <div style="
            display:flex;
            justify-content:space-between;
            width:75%;
            font-size:18px;
            font-weight:bold;
            margin-bottom:5px;
        ">
            <span>← Trust model more</span>
            <span>Trust observation more →</span>
        </div>
        """
    )
)

widgets.interact(
    show_assimilation,
    K=K_slider
)
```

```python
def update_model(model, observation, K=0.5):
    return model + K * (observation - model)


model_guess = 0.10
observation = 0.16

for K in [0.0, 0.25, 0.5, 0.75, 1.0]:
    updated = update_model(
        model_guess,
        observation,
        K
    )

    print(
        f"K={K:>4}: "
        f"updated chlorophyll = "
        f"{updated:.3f} mg m-3"
    )
```

```python
from sklearn.ensemble import RandomForestRegressor

# --------------------------------------------------
# Focus on the observed subsurface chlorophyll layer
# --------------------------------------------------

DEPTH_TOP = -80
DEPTH_BOTTOM = -140


# --------------------------------------------------
# BGC-Argo observations
# --------------------------------------------------

argo_time_ai = pd.to_datetime(
    ds["argo_time"].values
)

argo_depth_ai = ds[
    "argo_depth"
].values

argo_chl_ai = ds[
    "argo_chlorophyll"
].values


# --------------------------------------------------
# Keep only valid observations between 80 and 140 m
# --------------------------------------------------

valid = (
    np.isfinite(argo_chl_ai)
    &
    np.isfinite(argo_depth_ai)
    &
    (argo_depth_ai <= DEPTH_TOP)
    &
    (argo_depth_ai >= DEPTH_BOTTOM)
)

obs_time = argo_time_ai[valid]
obs_depth = argo_depth_ai[valid]
obs_chl = argo_chl_ai[valid]


# --------------------------------------------------
# Numerical-model grid
# --------------------------------------------------

model_times = pd.DatetimeIndex(
    pd.to_datetime(ds["time"].values)
)

model_depths = ds[
    "z"
].values

model_chl = ds[
    "total_chlorophyll"
].values


# --------------------------------------------------
# Fast matching to nearest numerical-model
# time and depth
# --------------------------------------------------

time_index = model_times.get_indexer(
    obs_time,
    method="nearest"
)

depth_index = np.abs(
    obs_depth[:, None]
    -
    model_depths[None, :]
).argmin(axis=1)

matched_model = model_chl[
    time_index,
    depth_index
]


# --------------------------------------------------
# Create BGC-Argo profile dates
# --------------------------------------------------

profile_date = pd.DatetimeIndex(
    obs_time
).normalize()

unique_dates = np.sort(
    profile_date.unique()
)


# --------------------------------------------------
# Split by WHOLE profiles
#
# Earlier 70% of profile dates = training
# Later 30% = testing
# --------------------------------------------------

n_train_dates = int(
    0.7 * len(unique_dates)
)

train_dates = unique_dates[
    :n_train_dates
]

test_dates = unique_dates[
    n_train_dates:
]

train_mask = np.isin(
    profile_date,
    train_dates
)

test_mask = np.isin(
    profile_date,
    test_dates
)


# --------------------------------------------------
# AI target:
# BGC-Argo - numerical model
# --------------------------------------------------

residual = (
    obs_chl
    -
    matched_model
)


# --------------------------------------------------
# Predictors
# --------------------------------------------------

day_of_year = (
    pd.DatetimeIndex(obs_time)
    .dayofyear
    .values
)

X = np.column_stack([
    matched_model,
    obs_depth,

    np.sin(
        2 * np.pi
        * day_of_year
        / 365.25
    ),

    np.cos(
        2 * np.pi
        * day_of_year
        / 365.25
    ),
])


# --------------------------------------------------
# Train a small Random Forest
# --------------------------------------------------

rf = RandomForestRegressor(
    n_estimators=50,
    max_depth=5,
    min_samples_leaf=4,
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X[train_mask],
    residual[train_mask]
)


# --------------------------------------------------
# Predict unseen later profiles
# --------------------------------------------------

predicted_residual = rf.predict(
    X[test_mask]
)

hybrid = (
    matched_model[test_mask]
    +
    predicted_residual
)


print(
    f"Training profiles : {len(train_dates)}"
)

print(
    f"Testing profiles  : {len(test_dates)}"
)

print(
    "✓ Subsurface BGC-Argo AI example complete"
)
```

```python
def rmse(a, b):
    return np.sqrt(
        np.nanmean(
            (
                np.asarray(a)
                -
                np.asarray(b)
            ) ** 2
        )
    )


rmse_model = rmse(
    matched_model[test_mask],
    obs_chl[test_mask]
)

rmse_hybrid = rmse(
    hybrid,
    obs_chl[test_mask]
)


improvement = (
    100
    * (
        rmse_model
        -
        rmse_hybrid
    )
    / rmse_model
)


print(
    f"Numerical-model RMSE : "
    f"{rmse_model:.4f} mg m-3"
)

print(
    f"Model + AI RMSE      : "
    f"{rmse_hybrid:.4f} mg m-3"
)

print(
    f"RMSE improvement     : "
    f"{improvement:.1f}%"
)
```

## 10. What have we learned?

We started with three incomplete views of one ocean system.

### <span style="color:blue;"><b>(1) Satellite</b></span>

Excellent view of surface chlorophyll, but little direct information about the deeper water column.

### <span style="color:blue;"><b>(2) BGC-Argo</b></span>

Excellent vertical information, but only along the float trajectory and at specific times.

### <span style="color:blue;"><b>(3) Numerical model</b></span>

Provides a continuous estimate through time and depth, but it is imperfect.

### <span style="color:blue;"><b>(4) Data Assimilation</b></span>

Uses observations to adjust the evolving numerical model.

### <span style="color:blue;"><b>(5) AI</b></span>

Can potentially learn systematic remaining errors, but must be used and tested carefully.

The important idea is therefore not:

> **observations OR models OR AI OR Data assimilation**

but:

> **observations + numerical models + data assimilation + carefully constrained AI**
